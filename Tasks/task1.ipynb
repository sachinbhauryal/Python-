```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
print("hello")
```

```python
df = pd.read_csv("data1.csv",skiprows=4)
```

```python
df.head()
```

```python
# taking the data from the latest year 
year = '2024'
```

```python
df[["Country Name",year]].head()
```

```python
print("keep only Country name and population :-")
population = df[["Country Name",year]].dropna()
print(population)
```

```python
print("Select by population and select top 10 companies :- ")
top10 = population.sort_values(by=year, ascending=False).head(10)
print(top10)
```

```python
# create a bar chart 

print("bar chart ")
plt.figure(figsize=(12,6))
plt.bar(top10["Country Name"],top10[year],color="steelblue")
plt.title("top 10 Most populated country in 2024")
plt.xlabel("Country")
plt.ylabel("population")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
population = df[year].dropna()

plt.figure(figsize=(10,6))
plt.hist(population,bins=30,edgecolor="black")
plt.title("Distribution of country population (2024)")
plt.xlabel("population")
plt.ylabel("number of Countries")
plt.tight_layout()
plt.show()
```
